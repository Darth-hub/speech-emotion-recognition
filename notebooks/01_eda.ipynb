# 01 — Exploratory Data Analysis

After running `python -m src.dataset` to build the manifest:
1. Class distribution per dataset
2. Speaker count and clips-per-speaker
3. Duration histogram
4. A few sample spectrograms per emotion


```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('../data/processed/manifest.csv')
df.groupby(['dataset','emotion']).size().unstack(fill_value=0)
```

```python
ax = df['emotion'].value_counts().plot(kind='bar', figsize=(8,4), title='Class distribution')
ax.set_ylabel('Clips'); plt.tight_layout(); plt.show()
```
